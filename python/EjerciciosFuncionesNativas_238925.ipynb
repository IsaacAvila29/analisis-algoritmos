## 1. Calcular la velocidad de un automovil

Solicitar la distancia recorrida y el tiempo utilizado. Calcular la velocidad del automovil.


```python
distancia = float(input("Distancia recorrida (km): "))
tiempo = float(input("Tiempo utilizado (h): "))

velocidad = distancia / tiempo
print(f"La velocidad es {velocidad:.2f} km/h")
```

## 2. Convertir grados Celsius a Fahrenheit

Solicitar una temperatura en grados Celsius y mostrar su equivalente en Fahrenheit.


```python
celsius = float(input("Temperatura en Celsius: "))

fahrenheit = celsius * 9 / 5 + 32
print(f"{celsius} C equivalen a {fahrenheit:.2f} F")
```

## 3. Calcular el area de un rectangulo

Pedir la base y la altura de un rectangulo y mostrar su area.


```python
base = float(input("Base: "))
altura = float(input("Altura: "))

area = base * altura
print(f"El area del rectangulo es {area:.2f}")
```

## 4. Calcular el promedio de tres calificaciones

Solicitar tres calificaciones, calcular el promedio y mostrar el resultado.


```python
cal1 = float(input("Calificacion 1: "))
cal2 = float(input("Calificacion 2: "))
cal3 = float(input("Calificacion 3: "))

promedio = (cal1 + cal2 + cal3) / 3
print(f"El promedio es {promedio:.2f}")
```

## 5. Calcular el area de un circulo

Solicitar el radio de un circulo y calcular su area.


```python
import math

radio = float(input("Radio del circulo: "))

area = math.pi * radio ** 2
print(f"El area del circulo es {area:.2f}")
```

## 6. Calcular el perimetro de un rectangulo

Solicitar la base y la altura de un rectangulo y calcular su perimetro.


```python
base = float(input("Base: "))
altura = float(input("Altura: "))

perimetro = 2 * (base + altura)
print(f"El perimetro del rectangulo es {perimetro:.2f}")
```

## 14. Calcular un descuento

Solicitar el precio de un producto y un porcentaje de descuento. Mostrar el descuento y el precio final.


```python
precio = float(input("Precio del producto: "))
porcentaje = float(input("Porcentaje de descuento: "))

descuento = precio * porcentaje / 100
precio_final = precio - descuento

print(f"Descuento: ${descuento:.2f}")
print(f"Precio final: ${precio_final:.2f}")
```
